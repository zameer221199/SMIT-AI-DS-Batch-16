# Question No. 07

```python
# for loop
num = 5
fact = 1  

for i in range(1, num + 1):  
    fact = fact * i        

print("Factorial:", fact)
```

```python
# for while loop
num = 5
fact = 1
i = 1 

while i <= num:     
    fact = fact * i  
    i = i + 1        

print("Factorial:", fact)
```
